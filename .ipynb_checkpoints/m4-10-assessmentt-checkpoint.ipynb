### Task 1 — Unsupervised Exploration

````python
# Import libraries for data handling
import numpy as np
import pandas as pd

# Import libraries for visualization
import matplotlib.pyplot as plt
import seaborn as sns

# Import preprocessing and dimensionality reduction tools
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE

# Import clustering algorithms
from sklearn.cluster import KMeans, DBSCAN

# Import clustering evaluation metrics
from sklearn.metrics import (
    silhouette_score,
    adjusted_rand_score,
    normalized_mutual_info_score
)

# Set a clean default style for plots
sns.set(style="whitegrid")

# Set default figure size
plt.rcParams["figure.figsize"] = (10, 6)

print("Libraries imported successfully.")
````

````python
# Load the Palmer Penguins dataset from seaborn
penguins = sns.load_dataset("penguins")

# Display the first rows of the dataset
display(penguins.head())

# Display the shape of the original dataset
print("Original dataset shape:", penguins.shape)

# Display general information about the dataset
print("\nDataset information:")
penguins.info()

# Check missing values in each column
print("\nMissing values in each column:")
print(penguins.isnull().sum())

# Display summary statistics for numerical columns
print("\nSummary statistics for numerical features:")
display(penguins.describe())
````

### Dataset Inspection

The Palmer Penguins dataset contains 344 rows and 7 columns.  
The columns include three categorical variables (`species`, `island`, and `sex`) and four numerical measurement variables: `bill_length_mm`, `bill_depth_mm`, `flipper_length_mm`, and `body_mass_g`.

The inspection shows that the dataset has a small number of missing values.  
The numerical measurement columns each contain 2 missing values, while the `sex` column contains 11 missing values. Since the unsupervised learning methods in this task require complete numerical input, these missing rows need to be removed before applying scaling, PCA, t-SNE, K-Means, and DBSCAN.

The summary statistics also show that the numerical features are measured on different scales. For example, `body_mass_g` ranges from 2700 to 6300, while `bill_depth_mm` ranges only from 13.1 to 21.5. This confirms that feature scaling is necessary before clustering, otherwise larger-scale variables such as body mass could dominate the distance calculations.

````python
# Remove rows with missing values
penguins_clean = penguins.dropna().copy()

# Check the shape after cleaning
print("Cleaned dataset shape:", penguins_clean.shape)

# Confirm that there are no missing values left
print("\nMissing values after cleaning:")
print(penguins_clean.isnull().sum())

# Select only numeric features for unsupervised learning
numeric_features = penguins_clean.select_dtypes(include=["float64", "int64"]).columns
X_numeric = penguins_clean[numeric_features]

# Store the true species labels only for later evaluation
true_species = penguins_clean["species"]

print("\nNumeric features used for clustering:")
print(list(numeric_features))
````

### Data Cleaning and Feature Selection

After removing rows with missing values, the dataset was reduced from 344 rows to 333 rows.  
This is still a sufficiently large dataset for the unsupervised exploration, and all remaining columns have complete values.

For clustering, only the four numerical features were selected: `bill_length_mm`, `bill_depth_mm`, `flipper_length_mm`, and `body_mass_g`.  
The actual `species` labels were not used during clustering. They were stored separately only for later evaluation, where the clustering results are compared with the real species labels using Adjusted Rand Score and Normalized Mutual Information.

````python
# Plot histograms of the numeric features to examine distributions
X_numeric.hist(figsize=(12, 8), bins=20)
plt.suptitle("Distributions of Numeric Features", fontsize=16)
plt.tight_layout()
plt.show()

# Plot boxplots to check possible anomalies or outliers
plt.figure(figsize=(12, 8))

for index, column in enumerate(numeric_features, 1):
    plt.subplot(2, 2, index)
    sns.boxplot(data=penguins_clean, y=column)
    plt.title(f"Boxplot of {column}")

plt.tight_layout()
plt.show()
````

### Distribution and Anomaly Check

The histograms show that the numerical features have natural biological variation.  
Some variables, such as `flipper_length_mm` and `body_mass_g`, show visible grouping patterns, which may be related to differences between penguin species.

The boxplots do not show extreme outliers that require manual removal.  
The minimum and maximum values are within realistic ranges for penguin body measurements, so the main cleaning step is removing missing values rather than deleting additional observations as anomalies.

This is important because removing valid biological variation could harm the clustering analysis. Therefore, all complete rows were kept after the missing-value cleaning step.

````python
# Scale numeric features so that all measurements have a similar range
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_numeric)

print("Scaled data shape:", X_scaled.shape)
````

````python
# Apply PCA with two components
pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X_scaled)

# Apply t-SNE with two components
tsne = TSNE(
    n_components=2,
    random_state=42,
    perplexity=30,
    learning_rate="auto",
    init="pca"
)
X_tsne = tsne.fit_transform(X_scaled)

# Print the explained variance ratio of PCA
print("PCA explained variance ratio:")
print(pca.explained_variance_ratio_)
````

````python
# Create a dataframe for visualization
visual_df = pd.DataFrame({
    "PCA1": X_pca[:, 0],
    "PCA2": X_pca[:, 1],
    "TSNE1": X_tsne[:, 0],
    "TSNE2": X_tsne[:, 1],
    "species": true_species.values
})

# Plot PCA and t-SNE projections side by side
plt.figure(figsize=(14, 6))

plt.subplot(1, 2, 1)
sns.scatterplot(
    data=visual_df,
    x="PCA1",
    y="PCA2",
    hue="species",
    s=70
)
plt.title("PCA Projection Colored by Actual Species")

plt.subplot(1, 2, 2)
sns.scatterplot(
    data=visual_df,
    x="TSNE1",
    y="TSNE2",
    hue="species",
    s=70
)
plt.title("t-SNE Projection Colored by Actual Species")

plt.tight_layout()
plt.show()
````

### PCA and t-SNE Visualization

The PCA plot shows a clear separation for the Gentoo species, which forms a distinct group on the right side of the projection.  
Adelie and Chinstrap penguins are closer to each other and show more overlap, which suggests that their numerical body measurements are more similar.

The t-SNE plot also shows a strong structure in the dataset.  
Gentoo is again clearly separated, while Adelie and Chinstrap appear closer but still form different local groups. This supports the idea that the dataset contains meaningful species-related patterns even before supervised learning is applied.

However, the overlap between Adelie and Chinstrap also suggests that unsupervised clustering may not perfectly recover all three species.  
This is because clustering algorithms only use measurement similarity and do not know the real biological class labels.

````python
# Apply K-Means clustering with k=3 because the dataset has three species
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
kmeans_labels = kmeans.fit_predict(X_scaled)

# Calculate evaluation metrics for K-Means
kmeans_silhouette = silhouette_score(X_scaled, kmeans_labels)
kmeans_ari = adjusted_rand_score(true_species, kmeans_labels)
kmeans_nmi = normalized_mutual_info_score(true_species, kmeans_labels)

print("K-Means results:")
print("Silhouette Score:", round(kmeans_silhouette, 4))
print("Adjusted Rand Score:", round(kmeans_ari, 4))
print("Normalized Mutual Information:", round(kmeans_nmi, 4))
````

### K-Means Clustering Results

K-Means was applied with `k=3` because the dataset contains three real penguin species.  
The model achieved a silhouette score of 0.4462, which indicates a moderate level of cluster separation.

The Adjusted Rand Score was 0.7994 and the Normalized Mutual Information score was 0.7899.  
These values show that K-Means recovered a large part of the real species structure, although the match was not perfect.

The result is reasonable because Gentoo is clearly separated in the PCA and t-SNE visualizations, while Adelie and Chinstrap overlap more strongly.  
This overlap likely explains most of the clustering errors.

The warning shown by scikit-learn is related to a known Windows/MKL behavior in K-Means.  
It does not invalidate the clustering result, but it can usually be avoided by setting the environment variable `OMP_NUM_THREADS=2`.

````python
# Define different DBSCAN parameter combinations
dbscan_configs = [
    {"eps": 0.7, "min_samples": 5},
    {"eps": 1.0, "min_samples": 5},
    {"eps": 1.3, "min_samples": 5}
]

dbscan_results = []

# Train and evaluate DBSCAN with each parameter combination
for config in dbscan_configs:
    dbscan = DBSCAN(
        eps=config["eps"],
        min_samples=config["min_samples"]
    )
    
    dbscan_labels = dbscan.fit_predict(X_scaled)
    
    # In DBSCAN, label -1 means noise
    number_of_clusters = len(set(dbscan_labels) - {-1})
    number_of_noise_points = list(dbscan_labels).count(-1)
    
    # Silhouette score requires at least two clusters
    if number_of_clusters >= 2:
        dbscan_silhouette = silhouette_score(X_scaled, dbscan_labels)
    else:
        dbscan_silhouette = np.nan
    
    dbscan_ari = adjusted_rand_score(true_species, dbscan_labels)
    dbscan_nmi = normalized_mutual_info_score(true_species, dbscan_labels)
    
    dbscan_results.append({
        "Algorithm": f"DBSCAN eps={config['eps']}, min_samples={config['min_samples']}",
        "Silhouette Score": dbscan_silhouette,
        "Adjusted Rand Score": dbscan_ari,
        "Normalized Mutual Information": dbscan_nmi,
        "Number of Clusters": number_of_clusters,
        "Noise Points": number_of_noise_points,
        "Labels": dbscan_labels
    })

# Display DBSCAN results without the label arrays
dbscan_results_table = pd.DataFrame([
    {key: value for key, value in result.items() if key != "Labels"}
    for result in dbscan_results
])

display(dbscan_results_table)
````

### DBSCAN Clustering Results

DBSCAN was tested with three different `eps` values while keeping `min_samples=5`.  
The results show that DBSCAN was sensitive to the chosen parameter values.

With `eps=0.7`, DBSCAN found 2 clusters and marked 17 points as noise.  
With `eps=1.0` and `eps=1.3`, it found 2 clusters and only 1 noise point. These two configurations produced the same evaluation scores.

The best DBSCAN silhouette score was 0.4669, which is slightly higher than the K-Means silhouette score.  
However, DBSCAN only found 2 clusters instead of 3. This means that although the clusters were reasonably compact, DBSCAN did not fully recover the three-species structure of the dataset.

This result suggests that DBSCAN separated the most distinct group well, likely Gentoo, but merged the more overlapping species into one broader cluster.

````python
# Create a K-Means result row
kmeans_result = {
    "Algorithm": "K-Means k=3",
    "Silhouette Score": kmeans_silhouette,
    "Adjusted Rand Score": kmeans_ari,
    "Normalized Mutual Information": kmeans_nmi,
    "Number of Clusters": 3,
    "Noise Points": 0
}

# Combine K-Means and DBSCAN results into one comparison table
comparison_table = pd.DataFrame(
    [kmeans_result] +
    [
        {key: value for key, value in result.items() if key != "Labels"}
        for result in dbscan_results
    ]
)

display(comparison_table)
````

````python
# Keep only DBSCAN results with a valid silhouette score
valid_dbscan_results = [
    result for result in dbscan_results
    if not np.isnan(result["Silhouette Score"])
]

# Select the best DBSCAN result based on silhouette score
best_dbscan = max(valid_dbscan_results, key=lambda result: result["Silhouette Score"])
best_dbscan_labels = best_dbscan["Labels"]

print("Best DBSCAN result:")
print("Algorithm:", best_dbscan["Algorithm"])
print("Silhouette Score:", round(best_dbscan["Silhouette Score"], 4))
print("Adjusted Rand Score:", round(best_dbscan["Adjusted Rand Score"], 4))
print("Normalized Mutual Information:", round(best_dbscan["Normalized Mutual Information"], 4))
print("Number of Clusters:", best_dbscan["Number of Clusters"])
print("Noise Points:", best_dbscan["Noise Points"])
````

### Comparison of Clustering Methods

The comparison table shows an important difference between internal and external clustering evaluation.

DBSCAN achieved the highest silhouette score with `eps=1.0` and `min_samples=5`, reaching 0.4669.  
However, it produced only 2 clusters, while the real dataset contains 3 species. Its Adjusted Rand Score was 0.6506 and its Normalized Mutual Information was 0.7600.

K-Means had a slightly lower silhouette score of 0.4462, but it achieved better agreement with the real species labels.  
Its Adjusted Rand Score was 0.7994 and its Normalized Mutual Information was 0.7899.

Therefore, K-Means is the better method for recovering the actual species structure in this task.  
DBSCAN produced compact clusters, but it failed to separate all three species because Adelie and Chinstrap are not clearly separated by density.

````python
# Create a dataframe for PCA comparison plots
comparison_df = pd.DataFrame({
    "PCA1": X_pca[:, 0],
    "PCA2": X_pca[:, 1],
    "Actual Species": true_species.values,
    "K-Means Cluster": kmeans_labels,
    "Best DBSCAN Cluster": best_dbscan_labels
})

# Plot actual species, K-Means clusters, and best DBSCAN clusters
plt.figure(figsize=(18, 5))

plt.subplot(1, 3, 1)
sns.scatterplot(
    data=comparison_df,
    x="PCA1",
    y="PCA2",
    hue="Actual Species",
    s=70
)
plt.title("Actual Species on PCA Projection")

plt.subplot(1, 3, 2)
sns.scatterplot(
    data=comparison_df,
    x="PCA1",
    y="PCA2",
    hue="K-Means Cluster",
    palette="Set1",
    s=70
)
plt.title("K-Means Clusters on PCA Projection")

plt.subplot(1, 3, 3)
sns.scatterplot(
    data=comparison_df,
    x="PCA1",
    y="PCA2",
    hue="Best DBSCAN Cluster",
    palette="Set1",
    s=70
)
plt.title("Best DBSCAN Clusters on PCA Projection")

plt.tight_layout()
plt.show()
````

### PCA-Based Cluster Comparison

The PCA comparison plot confirms the numerical evaluation results.  
The actual species plot shows that Gentoo is clearly separated, while Adelie and Chinstrap are closer together and partially overlap.

The K-Means plot shows three clusters, which makes it more similar to the real species structure.  
It separates Gentoo well and also attempts to divide the left-side region into two groups, roughly corresponding to Adelie and Chinstrap.

The DBSCAN plot shows only two main clusters and one noise point.  
This means DBSCAN identified the strongest separation in the data, but it did not divide the overlapping Adelie and Chinstrap region into two separate species groups.

Overall, the visualization supports the conclusion that K-Means gives a better species-level clustering result, while DBSCAN gives a simpler density-based separation.

### Task 2 — Supervised Model Pipeline

````python
# Import preprocessing and pipeline tools for Task 2
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline

# Import model selection tools
from sklearn.model_selection import StratifiedKFold, cross_validate, GridSearchCV

# Import supervised learning models
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC

print("Task 2 imports loaded successfully.")
````

````python
# Prepare the dataset for supervised classification
penguins_supervised = penguins.dropna().copy()

# Define the target and feature variables
target = "species"
X = penguins_supervised.drop(columns=[target])
y = penguins_supervised[target]

# Identify numeric and categorical features
numeric_features = X.select_dtypes(include=["float64", "int64"]).columns.tolist()
categorical_features = X.select_dtypes(include=["object", "category"]).columns.tolist()

print("Supervised dataset shape:", penguins_supervised.shape)
print("Feature matrix shape:", X.shape)
print("Target shape:", y.shape)

print("\nNumeric features:")
print(numeric_features)

print("\nCategorical features:")
print(categorical_features)

print("\nTarget classes:")
print(y.unique())
````

````python
# Create a preprocessing pipeline for numeric and categorical columns
preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numeric_features),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features)
    ]
)

print("Preprocessor created successfully.")
````

````python
# Define the models to compare
models = {
    "Logistic Regression": LogisticRegression(max_iter=1000, random_state=42),
    "Random Forest": RandomForestClassifier(random_state=42),
    "SVC": SVC(random_state=42)
}

# Define stratified 5-fold cross-validation
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# Define evaluation metrics
scoring = {
    "accuracy": "accuracy",
    "precision_macro": "precision_macro",
    "recall_macro": "recall_macro",
    "f1_macro": "f1_macro"
}

print("Models and cross-validation setup completed.")
````

````python
# Evaluate each model using a full pipeline and stratified 5-fold cross-validation
cv_results = []

for model_name, model in models.items():
    pipeline = Pipeline(
        steps=[
            ("preprocessor", preprocessor),
            ("classifier", model)
        ]
    )
    
    scores = cross_validate(
        pipeline,
        X,
        y,
        cv=cv,
        scoring=scoring,
        return_train_score=False
    )
    
    cv_results.append({
        "Model": model_name,
        "Accuracy": scores["test_accuracy"].mean(),
        "Precision (Macro)": scores["test_precision_macro"].mean(),
        "Recall (Macro)": scores["test_recall_macro"].mean(),
        "F1 (Macro)": scores["test_f1_macro"].mean()
    })

# Create a results table
cv_results_df = pd.DataFrame(cv_results)

# Sort models by F1 score
cv_results_df = cv_results_df.sort_values(by="F1 (Macro)", ascending=False)

display(cv_results_df)
````

````python
# Select the best model based on macro F1 score
best_model_name = cv_results_df.iloc[0]["Model"]
best_model_f1 = cv_results_df.iloc[0]["F1 (Macro)"]

print("Best model based on macro F1-score:", best_model_name)
print("Best macro F1-score:", round(best_model_f1, 4))
````

````python
# Create the default pipeline for the selected best model
best_default_model = models[best_model_name]

best_default_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("classifier", best_default_model)
    ]
)

print("Default best pipeline created for:", best_model_name)
````

````python
# Define a hyperparameter grid for the selected best model
if best_model_name == "Logistic Regression":
    param_grid = {
        "classifier__C": [0.01, 0.1, 1, 10],
        "classifier__solver": ["lbfgs", "saga"],
        "classifier__class_weight": [None, "balanced"]
    }

elif best_model_name == "Random Forest":
    param_grid = {
        "classifier__n_estimators": [50, 100, 200],
        "classifier__max_depth": [None, 3, 5, 10],
        "classifier__min_samples_split": [2, 5, 10]
    }

elif best_model_name == "SVC":
    param_grid = {
        "classifier__C": [0.1, 1, 10],
        "classifier__kernel": ["linear", "rbf", "poly"],
        "classifier__gamma": ["scale", "auto"]
    }

print("Hyperparameter grid for", best_model_name)
print(param_grid)
````

````python
# Run GridSearchCV using stratified 5-fold cross-validation
grid_search = GridSearchCV(
    estimator=best_default_pipeline,
    param_grid=param_grid,
    scoring="f1_macro",
    cv=cv,
    n_jobs=-1,
    verbose=1
)

grid_search.fit(X, y)

print("Grid search completed.")
print("Best parameters:")
print(grid_search.best_params_)

print("\nBest cross-validation F1 score:")
print(round(grid_search.best_score_, 4))
````

````python
# Compare the default best model with the tuned model
default_scores = cross_validate(
    best_default_pipeline,
    X,
    y,
    cv=cv,
    scoring=scoring,
    return_train_score=False
)

comparison_default_tuned = pd.DataFrame({
    "Model": [
        f"Default {best_model_name}",
        f"Tuned {best_model_name}"
    ],
    "Macro F1-score": [
        default_scores["test_f1_macro"].mean(),
        grid_search.best_score_
    ]
})

display(comparison_default_tuned)
````

## Task 2 Interpretation

In this task, a supervised machine learning pipeline was created to classify penguin species.  
The cleaned Palmer Penguins dataset contained 333 complete observations after removing rows with missing values.

The target variable was `species`, while the remaining columns were used as input features.  
The feature set included four numerical measurement columns: `bill_length_mm`, `bill_depth_mm`, `flipper_length_mm`, and `body_mass_g`. It also included two categorical columns: `island` and `sex`.

A preprocessing pipeline was built using `ColumnTransformer`.  
Numerical features were scaled with `StandardScaler`, while categorical features were encoded with `OneHotEncoder`. This allowed the models to use both numerical measurements and categorical information in one consistent pipeline.

Three classification models were compared using stratified 5-fold cross-validation: Logistic Regression, Random Forest, and SVC.  
The models were evaluated with accuracy, macro precision, macro recall, and macro F1-score. Macro F1-score was used as the main selection metric because it balances precision and recall across all three species.

Logistic Regression and SVC achieved the best default performance, both with an accuracy of 0.9940 and a macro F1-score of 0.9923.  
Random Forest also performed strongly, with an accuracy of 0.9910 and a macro F1-score of 0.9891. Since Logistic Regression and SVC were tied, Logistic Regression was selected as the best model by the pipeline.

GridSearchCV was then used to tune the Logistic Regression model.  
The best parameters were `C=10`, `class_weight='balanced'`, and `solver='liblinear'`. The tuned model achieved a macro F1-score of 0.9964, compared with 0.9923 for the default Logistic Regression model. This shows a small but positive improvement after hyperparameter tuning.

Overall, the supervised pipeline performed very strongly on the Palmer Penguins dataset.  
Compared with the unsupervised methods in Task 1, the supervised approach produced better species prediction performance because it learned directly from the real species labels and used both numerical and categorical features.

### Task 3 — Model Evaluation & Interpretation

````python
# Import additional tools needed for Task 3
from sklearn.model_selection import train_test_split, learning_curve
from sklearn.metrics import (
    classification_report,
    ConfusionMatrixDisplay,
    roc_curve,
    auc
)
from sklearn.preprocessing import label_binarize
from sklearn.inspection import permutation_importance
from sklearn.base import clone

print("Task 3 imports loaded successfully.")
````

````python
# Split the data into training and test sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training feature shape:", X_train.shape)
print("Test feature shape:", X_test.shape)
print("Training target shape:", y_train.shape)
print("Test target shape:", y_test.shape)

print("\nTraining class distribution:")
print(y_train.value_counts())

print("\nTest class distribution:")
print(y_test.value_counts())
````

````python
# Clone the best tuned model from GridSearchCV
best_tuned_model = clone(grid_search.best_estimator_)

# Train the tuned model on the training set
best_tuned_model.fit(X_train, y_train)

# Predict species on the test set
y_pred = best_tuned_model.predict(X_test)

print("Best tuned model trained successfully.")
print("Model used:", best_model_name)
print("Best parameters:", grid_search.best_params_)
````

````python
# Print the full classification report
print("Classification Report for Best Tuned Model")
print(classification_report(y_test, y_pred))
````

````python
# Plot the confusion matrix
ConfusionMatrixDisplay.from_estimator(
    best_tuned_model,
    X_test,
    y_test,
    cmap="Blues"
)

plt.title("Confusion Matrix for Best Tuned Model")
plt.show()
````

### Classification Report and Confusion Matrix

The tuned Logistic Regression model was evaluated on a held-out 20% test set containing 67 observations.  
The split was stratified, so the train and test sets preserved the species distribution.

The classification report shows very strong performance.  
The model achieved an overall accuracy of 0.99. Gentoo was classified perfectly with precision, recall, and F1-score equal to 1.00. Adelie also performed very well, with an F1-score of 0.98. Chinstrap had a slightly lower precision of 0.93, but its recall was 1.00, meaning all Chinstrap examples in the test set were correctly identified.

The confusion matrix confirms that the model made only one mistake.  
One Adelie penguin was predicted as Chinstrap, while all Chinstrap and Gentoo penguins were classified correctly.

This suggests that the hardest separation is between Adelie and Chinstrap, which is consistent with the PCA and t-SNE visualizations from Task 1 where these two species showed more overlap.

````python
# Get class names from the trained model
class_names = best_tuned_model.classes_

# Binarize the test labels for one-vs-rest ROC curves
y_test_binarized = label_binarize(y_test, classes=class_names)

# Get predicted probabilities
y_score = best_tuned_model.predict_proba(X_test)

# Plot ROC curves for each species
plt.figure(figsize=(8, 6))

for i, class_name in enumerate(class_names):
    fpr, tpr, _ = roc_curve(y_test_binarized[:, i], y_score[:, i])
    roc_auc = auc(fpr, tpr)
    
    plt.plot(
        fpr,
        tpr,
        label=f"{class_name} (AUC = {roc_auc:.3f})"
    )

# Add reference line for random classifier
plt.plot([0, 1], [0, 1], linestyle="--", label="Random Classifier")

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("One-vs-Rest ROC Curves")
plt.legend()
plt.grid(True)
plt.show()
````

### ROC Curve Analysis

The one-vs-rest ROC curves show that the model separates each species extremely well.  
All three species achieved an AUC of 1.000, which means the model ranked the true class above the other classes almost perfectly on the test set.

This result supports the classification report and confusion matrix.  
Even though there was one final label mistake between Adelie and Chinstrap, the predicted probabilities still show very strong class separation overall.

The ROC results also show that Gentoo is the easiest species to separate, which matches the earlier unsupervised visualizations where Gentoo formed the clearest group.

````python
# Compute learning curves using macro F1-score
train_sizes, train_scores, validation_scores = learning_curve(
    estimator=best_tuned_model,
    X=X_train,
    y=y_train,
    cv=cv,
    scoring="f1_macro",
    train_sizes=np.linspace(0.1, 1.0, 10),
    n_jobs=-1
)

# Calculate mean and standard deviation
train_scores_mean = train_scores.mean(axis=1)
train_scores_std = train_scores.std(axis=1)

validation_scores_mean = validation_scores.mean(axis=1)
validation_scores_std = validation_scores.std(axis=1)

# Plot learning curves
plt.figure(figsize=(9, 6))

plt.plot(train_sizes, train_scores_mean, marker="o", label="Training score")
plt.plot(train_sizes, validation_scores_mean, marker="o", label="Cross-validation score")

plt.fill_between(
    train_sizes,
    train_scores_mean - train_scores_std,
    train_scores_mean + train_scores_std,
    alpha=0.2
)

plt.fill_between(
    train_sizes,
    validation_scores_mean - validation_scores_std,
    validation_scores_mean + validation_scores_std,
    alpha=0.2
)

plt.xlabel("Training Set Size")
plt.ylabel("Macro F1-score")
plt.title("Learning Curve for Best Tuned Model")
plt.legend()
plt.grid(True)
plt.show()
````

### Learning Curve Analysis

The learning curve shows that the model has very high training and cross-validation scores.  
The training score stays close to 1.00, while the cross-validation score increases as more training data is used.

The gap between the training and validation curves is small, especially with larger training sizes.  
This suggests that the model is not strongly overfitting. It generalizes well to unseen data.

There is also no clear sign of underfitting, because both the training and validation scores are high.  
Overall, the learning curve suggests that the tuned Logistic Regression model is stable and suitable for this classification task.

````python
%%writefile app.py
from flask import Flask, request, jsonify
import joblib
import pandas as pd
import os

app = Flask(__name__)

MODEL_PATH = "penguin_species_pipeline.joblib"

if not os.path.exists(MODEL_PATH):
    raise FileNotFoundError(f"Model file not found: {MODEL_PATH}")

model = joblib.load(MODEL_PATH)

REQUIRED_FIELDS = [
    "island",
    "bill_length_mm",
    "bill_depth_mm",
    "flipper_length_mm",
    "body_mass_g",
    "sex"
]

VALID_ISLANDS = ["Torgersen", "Biscoe", "Dream"]
VALID_SEX_VALUES = ["Male", "Female"]


@app.route("/health", methods=["GET"])
def health():
    return jsonify({
        "status": "ok",
        "message": "Penguin species prediction API is running"
    })


@app.route("/predict", methods=["POST"])
def predict():
    try:
        data = request.get_json()

        if data is None:
            return jsonify({
                "error": "Invalid request. JSON body is required."
            }), 400

        missing_fields = [field for field in REQUIRED_FIELDS if field not in data]

        if missing_fields:
            return jsonify({
                "error": "Missing required fields.",
                "missing_fields": missing_fields
            }), 400

        if data["island"] not in VALID_ISLANDS:
            return jsonify({
                "error": "Invalid island value.",
                "valid_values": VALID_ISLANDS
            }), 400

        if data["sex"] not in VALID_SEX_VALUES:
            return jsonify({
                "error": "Invalid sex value.",
                "valid_values": VALID_SEX_VALUES
            }), 400

        numeric_fields = [
            "bill_length_mm",
            "bill_depth_mm",
            "flipper_length_mm",
            "body_mass_g"
        ]

        for field in numeric_fields:
            if not isinstance(data[field], (int, float)):
                return jsonify({
                    "error": f"Field '{field}' must be numeric."
                }), 400

        input_df = pd.DataFrame([{
            "island": data["island"],
            "bill_length_mm": data["bill_length_mm"],
            "bill_depth_mm": data["bill_depth_mm"],
            "flipper_length_mm": data["flipper_length_mm"],
            "body_mass_g": data["body_mass_g"],
            "sex": data["sex"]
        }])

        prediction = model.predict(input_df)[0]
        probabilities = model.predict_proba(input_df)[0]

        probability_dict = {
            class_name: float(probability)
            for class_name, probability in zip(model.classes_, probabilities)
        }

        return jsonify({
            "predicted_species": prediction,
            "class_probabilities": probability_dict
        })

    except Exception as error:
        return jsonify({
            "error": "Prediction failed.",
            "details": str(error)
        }), 500


if __name__ == "__main__":
    app.run(debug=True)
````

## API Documentation

The trained penguin species classification model is deployed using a Flask application in `app.py`.  
The model file is saved as `penguin_species_pipeline.joblib`.

### Health Endpoint

**Endpoint:** `/health`  
**Method:** `GET`

This endpoint checks whether the API is running.

Example response:

```json
{
  "message": "Penguin species prediction API is running",
  "status": "ok"
}
```

### Prediction Endpoint

**Endpoint:** `/predict`  
**Method:** `POST`

This endpoint accepts penguin measurement data as JSON and returns the predicted species with class probabilities.

Expected input format:

```json
{
  "island": "Torgersen",
  "bill_length_mm": 39.1,
  "bill_depth_mm": 18.7,
  "flipper_length_mm": 181.0,
  "body_mass_g": 3750.0,
  "sex": "Male"
}
```

Example successful response:

```json
{
  "predicted_species": "Adelie",
  "class_probabilities": {
    "Adelie": 0.9999480473466742,
    "Chinstrap": 0.00003559970126304496,
    "Gentoo": 0.000016352952062726028
  }
}
```

### Error Handling

If required fields are missing, the API returns a `400` response.

Example error response:

```json
{
  "error": "Missing required fields.",
  "missing_fields": [
    "bill_depth_mm",
    "flipper_length_mm",
    "body_mass_g",
    "sex"
  ]
}
```

If a numerical field has an invalid data type, the API also returns a `400` response.

Example error response:

```json
{
  "error": "Field 'bill_length_mm' must be numeric."
}
```
